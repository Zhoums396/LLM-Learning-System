```python
data['ques_knowledges'] = data['ques_knowledges'].astype(str)
fdata = data[~data['ques_knowledges'].str.contains(',', na=False)]

fdata
```

```python
fdata['ques_knowledges'] = fdata['ques_knowledges'].str.replace(r"[\[\]']", '', regex=True)
fdata['ques_knowledges']
```

```python
fdata = fdata.reset_index(drop=True)
fdata
```

```python
import re
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
le = LabelEncoder()
fdata['difficulty_encoded'] = le.fit_transform(fdata['ques_difficulty'])
```

```python
def clean_text(text):
    # 转小写
    text = text.lower()
    # 去除标点符号和数字，保留中文字符、字母和空格
    text = re.sub(r'[^a-zA-Z\u4e00-\u9fa5\s]', '', text)
    return text

fdata['ques_content_cleaned'] = fdata['ques_content'].apply(clean_text)
```

```python
import jieba
from sklearn.feature_extraction.text import TfidfVectorizer

# 加载中文停用词文件（示例，确保停用词文件存在）
def load_stopwords(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        stopwords = set(f.read().splitlines())
    return list(stopwords)  # 转换为列表

# 中文分词函数
def chinese_tokenizer(text):
    return list(jieba.cut(text))

# 加载停用词
stopwords = load_stopwords('C:/Users/Administrator/Desktop/CJEval/chinese_stopwords.txt')  # 你需要提供中文停用词的文件

# 创建TF-IDF向量化器
tfidf = TfidfVectorizer(stop_words=stopwords, max_features=5000, tokenizer=chinese_tokenizer)

# 计算TF-IDF矩阵
X_tfidf = tfidf.fit_transform(fdata['ques_content_cleaned'])

# 打印结果
print(X_tfidf.shape)
```
